```python
# This script runs a live camera feed to capture image data. w = 3840   h = 2160
# Press the 's' key to capture after running the script to save an image. Press 'q' to quit.

import cv2
import os

save_dir = "captures"
os.makedirs(save_dir, exist_ok=True)

cap = cv2.VideoCapture(1, cv2.CAP_DSHOW) # See3CAM on channel 1

cap.set(cv2.CAP_PROP_FRAME_WIDTH, 3840)
cap.set(cv2.CAP_PROP_FRAME_HEIGHT, 2160)

#cap.set(cv2.CAP_PROP_FPS, 15)
print("Width:", cap.get(cv2.CAP_PROP_FRAME_WIDTH))
print("Height:", cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
print("FPS:", cap.get(cv2.CAP_PROP_FPS))

count = 0

while True:
    ret, frame = cap.read()
    if not ret:
        print("Failed to grab frame")
        break

    #frame = cv2.rotate(frame, cv2.ROTATE_90_CLOCKWISE)
    h, w = frame.shape[:2]
    preview = cv2.resize(frame, (int(w*0.6), int(h*0.6) ) )
    cv2.imshow("Live Camera", preview)  # show live feed
    
    key = cv2.waitKey(1) & 0xFF

    if key == ord('s'):  # press 's' to save frame
        filename = os.path.join(save_dir, f"img_{count:03d}.jpg") # saves as img_xxx.jpg
        cv2.imwrite(filename, frame)
        print("Saved:", filename)
        count += 1

    if key == ord('q'):  # press 'q' to quit
        break

cap.release()
cv2.destroyAllWindows()
```
