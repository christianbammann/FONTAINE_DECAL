```python
board_w_in = 30
board_h_in = 46

border_in = 1
square_in = 2

squares_x = 14
squares_y = 22

svg = []

svg.append(f'<svg xmlns="http://www.w3.org/2000/svg" width="{board_w_in}in" height="{board_h_in}in" viewBox="0 0 {board_w_in} {board_h_in}">')
svg.append(f'<rect x="0" y="0" width="{board_w_in}" height="{board_h_in}" fill="white"/>')

for y in range(squares_y):
    for x in range(squares_x):
        if (x + y) % 2 == 0:
            rx = border_in + x * square_in
            ry = border_in + y * square_in
            svg.append(
                f'<rect x="{rx}" y="{ry}" width="{square_in}" height="{square_in}" fill="black"/>'
            )

svg.append('</svg>')

with open("checkerboard.svg", "w") as f:
    f.write("\n".join(svg))

print("checkerboard.svg written")
```

```python
import xml.etree.ElementTree as ET

tree = ET.parse("checkerboard.svg")
root = tree.getroot()

# Board size
W = float(root.attrib["width"].replace("in",""))
H = float(root.attrib["height"].replace("in",""))

rects = [e for e in root if e.tag.endswith("rect")]

# first rect is the white background
black = rects[1:]

xs = [float(r.attrib["x"]) for r in black]
ys = [float(r.attrib["y"]) for r in black]
ws = [float(r.attrib["width"]) for r in black]
hs = [float(r.attrib["height"]) for r in black]

min_x = min(xs)
min_y = min(ys)
max_x = max(x + w for x, w in zip(xs, ws))
max_y = max(y + h for y, h in zip(ys, hs))

left_border   = min_x
top_border    = min_y
right_border  = W - max_x
bottom_border = H - max_y

usable_w = max_x - min_x
usable_h = max_y - min_y

unique_x = sorted(set(xs))
unique_y = sorted(set(ys))

print("Board size (in):", W, "x", H)
print("Square size (in):", ws[0], "x", hs[0])
print("Squares across:", len(unique_x))
print("Squares down:", len(unique_y))
print("Left border (in):", left_border)
print("Right border (in):", right_border)
print("Top border (in):", top_border)
print("Bottom border (in):", bottom_border)
print("Usable checkerboard area (in):", usable_w, "x", usable_h)
```
