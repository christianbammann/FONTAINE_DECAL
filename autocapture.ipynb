```python
# This script runs a live camera feed to capture image data automatically. w = 1920   h = 1080
# Press 'q' to quit.

import cv2
import os
import time

save_dir = "autocaptures10" # save directory
os.makedirs(save_dir, exist_ok=True)

cap = cv2.VideoCapture(1, cv2.CAP_DSHOW) # See3CAM on channel 1
cap.set(cv2.CAP_PROP_FRAME_WIDTH, 1920) # set width
cap.set(cv2.CAP_PROP_FRAME_HEIGHT, 1080) # set height

print("Width:", cap.get(cv2.CAP_PROP_FRAME_WIDTH))
print("Height:", cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
print("FPS:", cap.get(cv2.CAP_PROP_FPS))

count = 0

max_images = 10 # max images
interval = 3  # seconds

last_time = time.time()

while count < max_images:
    ret, frame = cap.read()
    if not ret:
        print("Failed to grab frame")
        break

    h, w = frame.shape[:2]
    preview = cv2.resize(frame, (int(w*0.4), int(h*0.4) ) )
    cv2.imshow("Live Camera", preview)  # show live feed

    current_time = time.time()

    if current_time - last_time >= interval:
        filename = os.path.join(save_dir, f"img{count:03d}.jpg") # saves as img_xxx.jpg
        cv2.imwrite(filename, frame)
        print("Saved:", filename)
        count += 1
        last_time = current_time

    if cv2.waitKey(1) & 0xFF == ord('q'): # press 'q' to quit
        break

cap.release()
cv2.destroyAllWindows()
```
